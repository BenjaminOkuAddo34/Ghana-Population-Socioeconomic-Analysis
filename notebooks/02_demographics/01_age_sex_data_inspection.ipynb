```python
from pathlib import Path
import pandas as pd

def find_project_root():
    here = Path.cwd().resolve()
    for folder in [here, *here.parents]:
        if (folder / "data").exists() and (folder / "README.md").exists():
            return folder
    raise FileNotFoundError("Project root not found. Check that data/ and README.md are in the project folder.")

PROJECT_ROOT = find_project_root()
RAW_DATA_DIR = PROJECT_ROOT / "data" / "raw"
PROCESSED_DATA_DIR = PROJECT_ROOT / "data" / "processed"
PROCESSED_DATA_DIR.mkdir(parents=True, exist_ok=True)

RAW_DATA_DIR.exists()
```

```python
age_sex_file = RAW_DATA_DIR / "population_table2_locality2.csv"

raw = pd.read_csv(age_sex_file, header=None, encoding="utf-8-sig")
print(raw.shape)
raw.head(10)
```

```python
df = raw.iloc[3:].copy()          
df.columns = ["locality", "geography", "sex"] + raw.iloc[2, 3:].tolist()
df.head(10)
```

```python
df = df[df["sex"].notna()].copy()

df[["locality", "geography"]] = df[["locality", "geography"]].ffill()

print(df.shape)
print(df["locality"].unique())
df.head(10)
```

```python
age_cols = df.columns[3:]                      
df[age_cols] = df[age_cols].apply(pd.to_numeric)

print(df.dtypes.value_counts())
print("Missing values:", df[age_cols].isna().sum().sum())
```

```python
df["locality"] = df["locality"].replace(
    {"All Locality Types": "All", "Urban2": "Urban", "Rural2": "Rural"}
)
print(df["locality"].value_counts())

national_total = df[
    (df["geography"] == "Ghana") & (df["locality"] == "All") & (df["sex"] == "Both sexes")
]["All ages"].iloc[0]

print("National total:", national_total)
print("Matches 30,832,019:", national_total == 30_832_019)
```

```python
age_groups = df.columns[4:]
print("Age groups add up to All ages:", (df[age_groups].sum(axis=1) == df["All ages"]).all())

by_sex = df.pivot_table(index=["locality", "geography"], columns="sex", values="All ages")
print("Male + Female = Both sexes:", ((by_sex["Male"] + by_sex["Female"]) == by_sex["Both sexes"]).all())

by_locality = df[df["sex"] == "Both sexes"].pivot_table(index="geography", columns="locality", values="All ages")
print("Urban + Rural = All:", ((by_locality["Urban"] + by_locality["Rural"]) == by_locality["All"]).all())
```

```python
REGIONS = ["Western", "Central", "Greater Accra", "Volta", "Eastern", "Ashanti", "Western North", "Ahafo",
           "Bono", "Bono East", "Oti", "Northern", "Savannah", "North East", "Upper East", "Upper West"]

both_all = df[(df["sex"] == "Both sexes") & (df["locality"] == "All")]

# Regions
regions_df = both_all[both_all["geography"].isin(REGIONS)]
print("Regions found:", regions_df["geography"].nunique())
print("Regions sum to national total:", regions_df["All ages"].sum() == 30_832_019)

# Everything else (districts and sub-metropolitan areas)
others = both_all[~both_all["geography"].isin(["Ghana"] + REGIONS)]
print("Districts and sub-metros:", len(others))
print("These sum to national total:", others["All ages"].sum() == 30_832_019)

# Sub-metropolitan areas have a prefix like KMA- or TMA-
sub = [g for g in others["geography"] if g.startswith(("STMA-", "CCMA-", "AMA-", "TMA-", "KMA-"))]
print("Sub-metros:", len(sub))
sub
```

```python
SUB_TO_PARENT = {
    "STMA-Essikado-Ketan": "Sekondi Takoradi Metropolitan Area (STMA)",
    "STMA-Sekondi": "Sekondi Takoradi Metropolitan Area (STMA)",
    "STMA-Takoradi": "Sekondi Takoradi Metropolitan Area (STMA)",
    "CCMA-Cape Coast North": "Cape Cape Metropolitan Area (CCMA)",
    "CCMA-Cape Coast South": "Cape Cape Metropolitan Area (CCMA)",
    "AMA-Ablekuma South": "Accra Metropolitan Area (AMA)",
    "AMA-Ashiedu Keteke": "Accra Metropolitan Area (AMA)",
    "AMA-Okaikoi South": "Accra Metropolitan Area (AMA)",
    "TMA-Tema Central": "Tema Metropolitan Area (TMA)",
    "TMA-Tema East": "Tema Metropolitan Area (TMA)",
    "KMA-Bantama": "Kumasi Metropolitan Area (KMA)",
    "KMA-Manhyia North": "Kumasi Metropolitan Area (KMA)",
    "KMA-Manhyia South": "Kumasi Metropolitan Area (KMA)",
    "KMA-Nhyiaeso": "Kumasi Metropolitan Area (KMA)",
    "KMA-Subin": "Kumasi Metropolitan Area (KMA)",
    "TMA-Tamale Central": "Tamale Metropolitan Area (TMA)",
    "TMA-Tamale South": "Tamale Metropolitan Area (TMA)",
}

# All rows except Ghana and the 16 regions
areas = df[~df["geography"].isin(["Ghana"] + REGIONS)].copy()
areas["district"] = areas["geography"].replace(SUB_TO_PARENT)

# Add the sub-metros together (for every locality and sex)
number_cols = list(df.columns[3:])
districts_long = areas.groupby(["district", "locality", "sex"], as_index=False)[number_cols].sum()

print("Districts:", districts_long["district"].nunique())

check = districts_long[(districts_long["locality"] == "All") & (districts_long["sex"] == "Both sexes")]
print("Sum of districts matches national total:", check["All ages"].sum() == 30_832_019)

check[check["district"].str.contains("Kumasi")]
```

```python
# both_all is still in the original file order: a region row, then its districts
region_of = {}
current_region = None
for g in both_all["geography"]:
    if g in REGIONS:
        current_region = g
    elif g != "Ghana":
        region_of[SUB_TO_PARENT.get(g, g)] = current_region

districts_long["region"] = districts_long["district"].map(region_of)
print("Districts without a region:", districts_long["region"].isna().sum())

# Check: do the districts in each region add up to that region's total?
d_tot = districts_long[(districts_long["locality"] == "All") & (districts_long["sex"] == "Both sexes")]
by_region = d_tot.groupby("region")["All ages"].sum()
region_totals = regions_df.set_index("geography")["All ages"]
print("Districts sum to each region's total:", (by_region == region_totals.reindex(by_region.index)).all())

d_tot.groupby("region")["district"].count().sort_values(ascending=False)
```

```python
# regions_df holds the "Both sexes", "All localities" rows for the 16 regions (from Step 7)
reg = regions_df.set_index("geography").copy()

reg["population"] = reg["All ages"]
reg["pop_0_14"] = reg[["0-4", "5-9", "10-14"]].sum(axis=1)
reg["pop_65_plus"] = reg["65 years and older"]
reg["pop_15_64"] = reg["population"] - reg["pop_0_14"] - reg["pop_65_plus"]

reg["pct_0_14"] = reg["pop_0_14"] / reg["population"] * 100
reg["pct_15_64"] = reg["pop_15_64"] / reg["population"] * 100
reg["pct_65_plus"] = reg["pop_65_plus"] / reg["population"] * 100
reg["dependency_ratio"] = (reg["pop_0_14"] + reg["pop_65_plus"]) / reg["pop_15_64"] * 100

reg[["population", "pct_0_14", "pct_15_64", "pct_65_plus", "dependency_ratio"]].round(1).sort_values(
    "dependency_ratio", ascending=False
)
```
